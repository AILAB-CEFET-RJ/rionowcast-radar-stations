```python
from pathlib import Path

import pandas as pd


pasta_origem = Path(
    "/home/noemi/atmoseer/data/alertario/pluviometricos_all"
)

pasta_destino = Path(
    "/home/noemi/atmoseer/data/alertario/pluviometricos_parquet"
)

pasta_destino.mkdir(parents=True, exist_ok=True)

arquivos = sorted(pasta_origem.rglob("*.parquet"))

print("Quantidade de arquivos:", len(arquivos))

if not arquivos:
    raise ValueError("Nenhum arquivo encontrado na pasta de origem.")

for i, arquivo in enumerate(arquivos, start=1):
    destino = pasta_destino / arquivo.name

    if destino.exists():
        print(f"[{i}/{len(arquivos)}] Já convertido: {arquivo.name}")
        continue

    try:
        # Apesar da extensão .parquet, o conteúdo original é CSV.
        df = pd.read_csv(arquivo)

        # Salva efetivamente no formato Parquet.
        df.to_parquet(
            destino,
            index=False,
            compression="snappy",
        )

        tamanho_original = arquivo.stat().st_size
        tamanho_convertido = destino.stat().st_size

        reducao = (
            1 - tamanho_convertido / tamanho_original
        ) * 100

        print(
            f"[{i}/{len(arquivos)}] "
            f"{arquivo.name} | "
            f"linhas={len(df):,} | "
            f"redução={reducao:.1f}%"
        )

    except Exception as erro:
        print(
            f"[{i}/{len(arquivos)}] "
            f"Erro em {arquivo.name}: {erro}"
        )

print("\nConversão finalizada.")
print("Arquivos convertidos em:", pasta_destino)
```

### analisando anos disponiveis de todas as estaçoes

```python
from pathlib import Path

import pandas as pd


pasta = Path(
    "/home/noemi/atmoseer/data/alertario/pluviometricos_parquet"
)

arquivos = sorted(pasta.glob("*.parquet"))

print("Quantidade de arquivos:", len(arquivos))

dados = []

for i, arquivo in enumerate(arquivos, start=1):

    df = pd.read_parquet(
        arquivo,
        columns=[
            "estacao_id",
            "estacao",
            "dia_utc",
        ],
    )

    df["dia_utc"] = pd.to_datetime(
        df["dia_utc"],
        utc=True,
        errors="coerce",
    )

    df = df.dropna(
        subset=[
            "estacao_id",
            "dia_utc",
        ]
    )

    df["ano"] = df["dia_utc"].dt.year

    # Guarda apenas as combinações únicas
    # de estação e ano presentes neste arquivo.
    estacoes_anos = df[
        [
            "estacao_id",
            "estacao",
            "ano",
        ]
    ].drop_duplicates()

    dados.append(estacoes_anos)

    if i % 50 == 0 or i == len(arquivos):
        print(f"Arquivos analisados: {i}/{len(arquivos)}")

resultado = (
    pd.concat(
        dados,
        ignore_index=True,
    )
    .drop_duplicates()
    .sort_values(
        [
            "estacao_id",
            "ano",
        ]
    )
    .reset_index(drop=True)
)

# Lista das estações.
estacoes = (
    resultado[
        [
            "estacao_id",
            "estacao",
        ]
    ]
    .drop_duplicates()
    .sort_values("estacao_id")
    .reset_index(drop=True)
)

print("\nQuantidade de estações:", len(estacoes))

print("\nEstações:")
display(estacoes)

# Anos disponíveis para cada estação.
anos_por_estacao = (
    resultado.groupby(
        [
            "estacao_id",
            "estacao",
        ]
    )["ano"]
    .apply(
        lambda anos: sorted(
            anos.unique().tolist()
        )
    )
    .reset_index(
        name="anos_disponiveis"
    )
)

print("\nAnos disponíveis por estação:")
display(anos_por_estacao)

# Quantidade de estações disponíveis em cada ano.
estacoes_por_ano = (
    resultado.groupby("ano")["estacao_id"]
    .nunique()
    .reset_index(
        name="quantidade_estacoes"
    )
)

estacoes_por_ano = estacoes_por_ano[
    estacoes_por_ano["ano"].between(
        2012,
        2024,
    )
]

print("\nQuantidade de estações por ano, entre 2012 e 2024:")
display(estacoes_por_ano)
```

```python
# id 34 Sumaré [2016, 2017, 2018, 2019]
# id 100 Teste [2017, 2019, 2020]
```

```python
anos_necessarios = set(
    range(2012, 2025)
)

# Identifica as estações que possuem todos esses anos.
estacoes_periodo_completo = anos_por_estacao[
    anos_por_estacao["anos_disponiveis"].apply(
        lambda anos: anos_necessarios.issubset(
            set(anos)
        )
    )
].copy()

print(
    "Quantidade de estações com dados de 2012 a 2024:",
    len(estacoes_periodo_completo),
)

print("\nEstações:")

display(
    estacoes_periodo_completo[
        [
            "estacao_id",
            "estacao",
            "anos_disponiveis",
        ]
    ]
)
```

```python
anos_por_estacao["anos_faltantes"] = (
    anos_por_estacao["anos_disponiveis"].apply(
        lambda anos: sorted(
            anos_necessarios - set(anos)
        )
    )
)

estacoes_periodo_incompleto = anos_por_estacao[
    anos_por_estacao["anos_faltantes"].apply(len) > 0
]

print(
    "Quantidade de estações sem todos os anos:",
    len(estacoes_periodo_incompleto),
)

display(
    estacoes_periodo_incompleto[
        [
            "estacao_id",
            "estacao",
            "anos_faltantes",
        ]
    ]
)
```
